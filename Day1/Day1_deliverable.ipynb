```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')
```

```python
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns
categorical_columns = df.select_dtypes(include=['object']).columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.drop(
    columns=['EmployeeCount', 'EmployeeNumber', 'Over18', 'StandardHours'],
    inplace=True
)
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(y=df['MonthlyIncome'])
plt.title('Box Plot of Monthly Income')
plt.ylabel('Monthly Income')
plt.show()
```

```python
Q1 = df['MonthlyIncome'].quantile(0.25)
Q3 = df['MonthlyIncome'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df['MonthlyIncome'] = df['MonthlyIncome'].clip(
    lower=lower_limit,
    upper=upper_limit
)
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Attrition')
plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x='Age', bins=20, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='Attrition', y='MonthlyIncome')
plt.title('Monthly Income by Attrition')
plt.xlabel('Attrition')
plt.ylabel('Monthly Income')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='OverTime', hue='Attrition')
plt.title('Attrition Based on Overtime')
plt.xlabel('Overtime')
plt.ylabel('Number of Employees')
plt.show()
```

```python
correlation_columns = [
    'Age',
    'MonthlyIncome',
    'TotalWorkingYears',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'JobLevel'
]

correlation = df[correlation_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
df['OverallSatisfaction'] = (
    df['JobSatisfaction']
    + df['EnvironmentSatisfaction']
    + df['RelationshipSatisfaction']
    + df['WorkLifeBalance']
) / 4
```

```python
df['PreviousExperienceYears'] = (
    df['TotalWorkingYears'] - df['YearsAtCompany']
)

df['PreviousExperienceYears'] = (
    df['PreviousExperienceYears'].clip(lower=0)
)
```

```python
df['OverTimeFlag'] = df['OverTime'].map({
    'Yes': 1,
    'No': 0
})
```

```python
df[
    [
        'OverallSatisfaction',
        'PreviousExperienceYears',
        'OverTimeFlag'
    ]
].head()
```

## Key Insights

1. Most employees stayed with the company, so the Attrition column is imbalanced.

2. Employees who work overtime appear to have a higher chance of leaving.

3. Employees who leave generally have lower monthly income than employees who stay.

4. Younger employees appear more frequently in the attrition group.

5. Total working years, job level, monthly income and years at the company show noticeable correlations.
